```python
parsed_logs = []

with open("day2_machine_log.txt", "r") as file:
    for line in file:
        # Skip empty lines if any exist
        if not line.strip():
            continue
            
        parts = line.split("|")
        
        log_dict = {
            "timestamp": parts[0].strip(),
            "line": parts[1].strip(),
            "level": parts[2].strip(),
            "message": parts[3].strip()
        }
        parsed_logs.append(log_dict)

print(f"Total logs parsed: {len(parsed_logs)}")
```

```python
from collections import Counter

def count_by_level(records):
    return dict(Counter(record["level"] for record in records))

# Example usage:
level_summary = count_by_level(parsed_logs)
print(level_summary)
```

```python
def errors_per_line(records):
    counts = {}
    
    for record in records:
        if record["level"] == "ERROR":
            line = record["line"]
            counts[line] = counts.get(line, 0) + 1
            
    if not counts:
        return None, 0
        
    worst_line = max(counts, key=counts.get)
    return worst_line, counts[worst_line]

# Example usage:
line_name, total_errors = errors_per_line(parsed_logs)
print(f"{line_name} had the most errors ({total_errors} total).")
```
