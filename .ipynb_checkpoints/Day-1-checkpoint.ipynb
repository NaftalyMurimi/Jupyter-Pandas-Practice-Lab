```python
import pandas as pd
import matplotlib.pyplot as plt

# Import data
data = pd.read_csv("C:/Users/User/Desktop/Kwidzyn Learn/jupyter_7day_practice_pack/day1/day1_sales.csv")
data.head()
```

```python
data.shape
```

```python
data.dtypes
```

```python
# Multiplies Price by Quantity to create a Total_Revenue column
data['Revenue'] = data['unit_price'] * data['units_sold']
data.head()
```

```python
revenue_per_region = data.groupby('region')['Revenue'].sum() 
revenue_per_region
```

```python
best_selling_product = data.groupby('product')['units_sold'].sum()
best_selling_product
```

```python
import matplotlib.pyplot as plt

# 1. Group the data first


# 2. Plot directly from Pandas
revenue_per_region.plot(kind='bar', color='blue', edgecolor='black')

# 3. Add required titles and labels
plt.title('Total Revenue by region')
plt.xlabel('Region')
plt.ylabel('Total Revenue ($)')
plt.xticks(rotation=45) # Tilts labels so they don't overlap

# 4. Display the chart cleanly
plt.tight_layout()
plt.show()
```
